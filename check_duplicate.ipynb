```python
from google.colab import drive
import pandas as pd

# ==== MOUNT GOOGLE DRIVE ====
drive.mount('/content/drive')

# ==== SESUAIKAN PATH FILE DI SINI (path di dalam Google Drive) ====
train_path = "/content/drive/MyDrive/Colab Datasets/BIO Splits v2/train_bio.csv"
val_path   = "/content/drive/MyDrive/Colab Datasets/BIO Splits v2/val_bio.csv"
test_path  = "/content/drive/MyDrive/Colab Datasets/BIO Splits v2/test_bio.csv"

# ==== LOAD DATA ====
df_train = pd.read_csv(train_path)
df_val   = pd.read_csv(val_path)
df_test  = pd.read_csv(test_path)

datasets = {
    "Train": df_train,
    "Val": df_val,
    "Test": df_test
}

print("="*60)
print("CEK DUPLIKAT BERDASARKAN KOLOM 'text'")
print("="*60)

# 1. Duplikat DI DALAM masing-masing split
print("\n[1] Duplikat internal (dalam satu split)")
for name, df in datasets.items():
    n_dup = df.duplicated(subset=["text"]).sum()
    print(f"  - {name}: {n_dup} baris duplikat dari total {len(df)} baris")

# 2. Overlap ANTAR split (potensi data leakage train->val/test)
print("\n[2] Overlap antar split (potensi data leakage)")
pairs = [("Train", "Val"), ("Train", "Test"), ("Val", "Test")]
for name_a, name_b in pairs:
    set_a = set(datasets[name_a]["text"])
    set_b = set(datasets[name_b]["text"])
    overlap = set_a & set_b
    print(f"  - {name_a} vs {name_b}: {len(overlap)} teks yang sama persis")

# 3. Detail baris duplikat internal (buat diperiksa manual)
print("\n[3] Detail duplikat internal per split")
for name, df in datasets.items():
    dup_rows = df[df.duplicated(subset=["text"], keep=False)].sort_values("text")
    if len(dup_rows) > 0:
        print(f"\n--- {name}: {len(dup_rows)} baris terlibat duplikat ---")
        display(dup_rows[["text"]].head(20))
    else:
        print(f"\n--- {name}: tidak ada duplikat ---")
```
