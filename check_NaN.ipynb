## **Check NaN**

```python
import pandas as pd

# Load the dataset
df = pd.read_csv(PATH)

# Display basic info about the dataframe
print("Dataset shape:", df.shape)
print("\nColumn names:", df.columns.tolist())

# Check for missing values per column
print("\nMissing values per column:")
print(df.isnull().sum())

# Optional: Show percentage of missing values per column
missing_percent = (df.isnull().sum() / len(df)) * 100
print("\nPercentage of missing values per column:")
print(missing_percent.round(2))

# Total missing values in the entire dataset
total_missing = df.isnull().sum().sum()
print(f"\nTotal missing values in the dataset: {total_missing}")

# Check if there is any missing value at all
if total_missing > 0:
    print("\n⚠️ The dataset contains missing values.")
    # Optional: show rows that have any missing value
    rows_with_na = df[df.isnull().any(axis=1)]
    print(f"\nNumber of rows with at least one missing value: {len(rows_with_na)}")
    print("First 5 rows with missing values:")
    print(rows_with_na.head())
else:
    print("\n✅ No missing values found in the dataset.")
```
